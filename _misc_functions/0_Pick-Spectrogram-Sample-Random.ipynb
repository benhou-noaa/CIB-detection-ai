```python
def grab_spectrogram_sample(path_to_folder, spectrogram_type, copy_destination):


    spectrogram_dict = {"B-DB":"*_B-DB*",
                        "B-S1K":"*_B-S1K*", 
                        "F-DB":"*_F-DB*", 
                        "F-ML":"*_F-ML237*", 
                        "F-S1K": "*_F-S1K*", 
                        "N": "*_N*"}
    
    all_spectrograms = [os.path.join(path_to_folder,i) for i in glob.glob(os.path.join(path_to_folder, spectrogram_dict[spectrogram_type]))]
    all_spectrograms = np.asarray(all_spectrograms)
    
    
    sample_size = 200
    if len(all_spectrograms) < sample_size:
        print("insufficient spectrograms to draw sample size of {size}, grabbing all spectrograms".format(size = sample_size))
        sample_size = len(all_spectrograms)
    
    

    sample_index = sorted(random.sample(range(len(all_spectrograms)), sample_size))
    spectrogram_sample = [all_spectrograms[i] for i in sample_index]
    print(len(spectrogram_sample))
    
    
    #create a subfolder for that type
    destination_path = os.path.join(copy_destination, spectrogram_type)
    
    if not os.path.exists(destination_path):
        os.mkdir(destination_path)
    print("made directory", destination_path, " as it did not exist")
    
    print("Now copying {sample_size} {spectrogram_type} to {destination}".format(sample_size=sample_size, 
                                                                                 spectrogram_type = spectrogram_type, 
                                                                                 destination = destination_path))
    for i in spectrogram_sample:
         
        shutil.copy(i, os.path.join(destination_path, os.path.basename(i)))
    


#rab_spectrogram_sample(path_to_folder_BF, "B-DB", output_directory)
#rab_spectrogram_sample(path_to_folder_BF, "B-S1K", output_directory)
#rab_spectrogram_sample(path_to_folder_BF, "F-DB", output_directory)
#rab_spectrogram_sample(path_to_folder_BF, "F-ML", output_directory)
grab_spectrogram_sample(path_to_folder_BF, "F-S1K", output_directory)
#rab_spectrogram_sample(path_to_folder_N, "N", output_directory)
```
